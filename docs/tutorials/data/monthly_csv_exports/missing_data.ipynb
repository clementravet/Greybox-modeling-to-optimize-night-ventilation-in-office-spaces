```python
from pathlib import Path
import pandas as pd


files = sorted(Path(".").glob("2025-0*.csv"))


if not files:
    print("No files matched pattern: 2025-0*.csv")
else:
    summary_rows = []

    for file in files:
        df = pd.read_csv(file)
        missing_per_col = df.isna().sum()
        total_missing = int(missing_per_col.sum())
        rows_with_missing = int(df.isna().any(axis=1).sum())

        print(f"\n{file.name}")
        print("-" * len(file.name))
        print(f"Total missing values: {total_missing}")
        print(f"Rows with at least one missing value: {rows_with_missing}")

        if total_missing > 0:
            print("Missing values by column:")
            print(missing_per_col[missing_per_col > 0].sort_values(ascending=False).to_string())

            # ── NEW: show line numbers ──────────────────────────────────────
            print("\nLines with missing values:")
            missing_mask = df.isna().any(axis=1)
            for idx, row in df[missing_mask].iterrows():
                line_num = idx + 2  # +1 for 0-based index, +1 for header row
                missing_cols = row.index[row.isna()].tolist()
                print(f"  Line {line_num}: {', '.join(missing_cols)}")
            # ──────────────────────────────────────────────────────────────
        else:
            print("No missing data found.")

        summary_rows.append(
            {
                "file": file.name,
                "total_missing_values": total_missing,
                "rows_with_missing": rows_with_missing,
            }
        )

    summary_df = pd.DataFrame(summary_rows).sort_values("file")
    print("\nSummary across files:")
    print(summary_df.to_string(index=False))
```
