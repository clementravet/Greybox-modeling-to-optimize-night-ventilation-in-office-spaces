```python
import pandas as pd
from pathlib import Path

input_file = "2025_full_hourly_R05.csv"
output_dir = Path(".")

df = pd.read_csv(input_file)

# Try to find a datetime column automatically
datetime_col = None
for col in df.columns:
    parsed = pd.to_datetime(df[col], errors="coerce")
    if parsed.notna().mean() > 0.8:  # at least 80% parseable as datetime
        datetime_col = col
        df[col] = parsed
        break

if datetime_col is None:
    raise ValueError("No datetime-like column found. Please specify the datetime column manually.")

# Export monthly files from April to October 2025
for month in range(4, 11):
    mask = (df[datetime_col].dt.year == 2025) & (df[datetime_col].dt.month == month)
    monthly_df = df.loc[mask].copy()
    output_file = output_dir / f"2025-{month:02d}_R05.csv"
    monthly_df.to_csv(output_file, index=False)
    print(f"Saved: {output_file} ({len(monthly_df)} rows)")
```
